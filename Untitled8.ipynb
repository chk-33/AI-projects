```python
def keyword_search(query):
    terms = query.lower().split()
    results = []

    for document in hybrid_documents:
        text = document.lower()
        matches = sum(term in text for term in terms)

        if matches > 0:
            results.append({
                "document": document,
                "keyword_matches": matches
            })

    return sorted(
        results,
        key=lambda item: item["keyword_matches"],
        reverse=True
    )
query = "What is the battery capacity of ZX-204?"

print("SEMANTIC SEARCH")
for result in semantic_search(query, top_k=3):
    print(round(result["score"], 4), result["document"])

print("\nKEYWORD SEARCH")
for result in keyword_search(query):
    print(result["keyword_matches"], result["document"])
```
